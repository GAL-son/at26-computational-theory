## 3.10
Find examples of subsets that are independent and dependent from the new decision table

```python
from itertools import combinations

C = table.get_conditionals()
print(f"Conditionals: {displayAsSet(C)}\n")

independence = []
for r in range(1, len(C) + 1):
    subsets = list(combinations(C, r))
    for subset in subsets:
        is_independent = table.is_subset_independent(list(subset))
        independence.append((subset, is_independent))
        print(f"{"Independent" if is_independent else "Dependent"} - {displayAsSet(subset)}")
```

**Code below allowes to interactively explore dependency of all subsets**

```python
from ipywidgets import interact, widgets, VBox, HBox
from IPython.display import display, clear_output
from itertools import combinations

choice = widgets.BoundedIntText(
    value=3,
    min=1,
    max=len(table.get_conditionals()),
    step=1,
    description='Subset length:',
    style={'description_width': 'initial'}
)

filter_dropdown = widgets.Dropdown(
    options=[('All', 'all'), ('Independent', 'independent'), ('Dependent', 'dependent')],
    value='all',
    description='Filter:',
    style={'description_width': 'initial'}
)

output = widgets.Output()

# 2. Wspólna funkcja aktualizująca
def update_results(change=None):
    length = choice.value
    filter_type = filter_dropdown.value
    
    with output:
        clear_output()
        print(f"--- Searching for length {length} (Filter: {filter_type}) ---")
        
        found_any = False
        for result in independence:
            subset, is_independent = result
            
            if len(subset) != length:
                continue

            if filter_type == 'independent' and not is_independent:
                continue

            if filter_type == 'dependent' and is_independent:
                continue

            label = "Independent" if is_independent else "Dependent"
            print(f"{label} - {displayAsSet(subset)}")
            found_any = True
        
        if not found_any:
            print("No matching subsets found.")

choice.observe(update_results, names='value')
filter_dropdown.observe(update_results, names='value')

display(VBox([choice, filter_dropdown]), output)
update_results()
```

## 3.12
Find examples of attribute subsets that are recuctions and conditional reductions of given decision table

## 3.12.1

```python
reducts = {}

for result in independence:
    subset, is_independent = result
    if is_independent:
        continue

    subset_reducts = table.get_attribute_reducts(list(subset))
    for reduct in subset_reducts:
        reducts[hash(str(reduct))] = (reduct, subset)
    
print("Reducts found:")
for reduct in reducts.values():
    print(f"{displayAsSet(reduct[0])} from subset {displayAsSet(reduct[1])}")    
```

## 3.12.2

```python
relative_reducts = {}

for result in conditional_independence:
    subset, is_independent = result
    if is_independent:
        continue

    subset_reducts = table.get_relative_reducts(family_D, list(subset))
    for reduct in subset_reducts:
        if not reduct:
            continue
        relative_reducts[hash(str(reduct))] = (reduct, subset)

print("Relative reducts found:")
for reduct in relative_reducts.values():
    print(f"{displayAsSet(reduct[0])} (from subset {displayAsSet(reduct[1])})")
```
