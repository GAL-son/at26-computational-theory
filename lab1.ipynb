### 3.1.1

Given data in 'dogs_1.csv' display qualities of this information system. This data includes:
* Value sets of the universe `U`
* Set of all atrubute collumns `A`
* Value sets `VAn` of all the atribute collumns `A`

```python
from src.aux import displayAsSet
from src.information_system import InformationSystem

system = InformationSystem(df, 'Obiekt')

print(f"U: {displayAsSet(system.get_universe())}")
print(f"A: {displayAsSet(system.get_attributes())}")
for i, attribute in enumerate(system.get_attributes()):
    print(f"VA{i}: {attribute}: {displayAsSet(system.get_attribute_values(attribute))}")
```

### 3.1.2

Given the previously defined information system, separate the attribute set into two sets:
* Single element decision set `D`
* Set of contitional elements `C`

and create decision table based on those

```python
from src.decision_table import DecisionTable

table = DecisionTable(df, 'Obiekt', 'Rasa')

# Enable caching to speed up repeated calls
table.enable_cache()

print(f"U: {displayAsSet(table.get_universe())}")
print(f"C: {displayAsSet(table.get_conditionals())}")
print(f"D: {displayAsSet(table.get_decision_attribute())}")
```

### 3.2
Given above decision table split objects into Equivalence classes based on the C-indiscernibility relation

```python
equivalence_classes = table.get_equivalence_classes()
for i, eq_class in enumerate(equivalence_classes):
    for obj in eq_class:
        print(f"[{obj}]C = ", end="")
    print(displayAsSet(eq_class))
print(f"C*: {equivalence_classes}")
```

### 3.3
Divide space U into subsets based on the race column and calculate upper and lower approximation of given sets

```python
subsets = {}
for value in table.get_decision_values():
    subsets[f"X{value[0:1].capitalize()}"] = table.get_objects_with_decision_value(value)

print("Subspaces of U based on decision attribute values:")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(subset)}")

print("\nLower approximations (C):")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(table.get_lower_approximation(subset))}")

print("\nUpper approximations: (C)")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(table.get_upper_approximation(subset))}")

```

### 3.4
Calculate C-positive, C-boundry, and C-negative regions of the subspaces from previous excercise

```python
print("Positive regions of subspaces U")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(table.get_positive_region(subset))}")

print("\nBoundry regions of subspaces U")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(table.get_boundary_region(subset))}")

print("\nNegative regions of subspaces U")
for value, subset in subsets.items():
    print(f"{value}: {displayAsSet(table.get_negative_region(subset))}")
```
